```python
import os
import shutil

home = os.getcwd()

#for jpg files
for filename in os.listdir(home):
    file_name, file_extension = os.path.splitext(filename)
    
    if file_extension.lower() == '.jpg':
        folder_name = f'{file_extension.replace(".", "")} files' 
        image_folder = os.path.join(home, folder_name)
        if not os.path.exists(image_folder):
            os.mkdir(image_folder)
            print(f'{folder_name} folder has been created...')
            
        old_image_path = os.path.join(home, filename)
        new_image_path = os.path.join(image_folder, filename)
        
        if old_image_path != new_image_path: 
            shutil.move(old_image_path, new_image_path)
            print(f'{filename} has been moved...')
```
